```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pickle
```

```python
#############FIX Me#########################
merge_path = "/centaurus/project/Playground/CT_Psoriasis/INN/psoriasis/result/shap_res"
##############################################

path=os.path.join(merge_path, f"value_merge_corrected.pickle")
with open(path,'rb') as f:
    shap_dict = pickle.load(f)
```

```python
# 여러개 그리기
for i in range(1, 7)[::-1]:    
    shap_dict[i]["mean"]=shap_dict[i].iloc[:,1:].mean(axis=1)
    shap_dict[i]=shap_dict[i].sort_values(by="mean",ascending=False)
    fig_input=shap_dict[i].iloc[0:5,:-1]
    fig_input.index=fig_input["name"]
    fig_input.drop("name",axis=1,inplace=True)
    fig_input=fig_input.T
    sns.reset_defaults()
    plt.clf()
    plt.figure(figsize=(8,6))
    ax = sns.barplot(data=fig_input,orient="h",errwidth=0)
    # plt.savefig(os.path.join(merge_path,f"shap_value_{i}_corrected.png"),bbox_inches="tight")
    plt.show()
```
